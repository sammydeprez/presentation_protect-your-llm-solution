```python
import textwrap
from langchain_openai import AzureChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

openai_completion_model = "gpt-35-turbo-open"
llm = AzureChatOpenAI(azure_deployment=openai_completion_model, api_version="2024-06-01")
```

```python
def print_messages(messages):
    for x in messages:
        lines = textwrap.wrap(x.pretty_repr(), width=80)
        for l in lines:
            print(l)
```

# Prompt Injection

```python
messages = []
messages.append(SystemMessage(content = "You help me only with questions related to swimming"))
messages.append(HumanMessage(content = "What is the best way to learn to swim"))

try:
    messages.append(llm.invoke(messages))
except Exception as e:
    messages.append(SystemMessage(content = f"{e.args[0]}"))

print_messages(messages)
```
